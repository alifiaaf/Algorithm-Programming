```python
def nilai(huruf, angka=0):
    if huruf == "A":
        return 4
    elif huruf == "B":
        return 3
    elif huruf == "C":
        return 2
    elif huruf == "D":
        return 1
    else:
        return angka

total = 0
jumlah = 0

while True:
    huruf = input("masukkan nilai: ").upper()

    if huruf == "":
        break

    hasil = nilai(huruf)
    print("nilai =", hasil)

    total = total + hasil
    jumlah = jumlah + 1

if jumlah > 0:
    rata = total / jumlah
    print("Rata-ratanya adalah:", rata)
```

```python
def tahun_kabisat(tahun):
    if tahun % 400 == 0:
        return True
    elif tahun % 100 == 0:
        return False
    elif tahun % 4 == 0:
        return True
    else:
        return False


def jumlah_hari(bulan, tahun):
    if bulan == 1 or bulan == 3 or bulan == 5 or bulan == 7 or bulan == 8 or bulan == 10 or bulan == 12:
        return 31
    elif bulan == 4 or bulan == 6 or bulan == 9 or bulan == 11:
        return 30
    elif bulan == 2:
        if tahun_kabisat(tahun):
            return 29
        else:
            return 28
    else:
        return 0

print("Program ini akan menentukan jumlah hari dalam bulan tertentu")

while True:
    print("Masukkan 0 untuk menghentikan program")
    bulan = int(input("masukkan bulan (1-12): "))

    if bulan == 0:
        print("Terima kasih sudah menggunakan program saya, sampai berjumpa lagi.")
        break

    tahun = 0

    if bulan == 2:
        tahun = int(input("masukkan tahun (e.g., 2021): "))

    hari = jumlah_hari(bulan, tahun)

    if hari == 0:
        print("Bulan tidak valid")
    else:
        print(hari, "hari dalam sebulan")
```
