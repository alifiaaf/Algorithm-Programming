```python
def cek_prima(angka):
    if angka <= 1:
        return False

    for i in range(2, angka):
        if angka % i == 0:
            return False

    return True

def tampilkan_hasil(angka):
    if cek_prima(angka):
        print(angka, "adalah bilangan Prima")
    else:
        print(angka, "bukanlah bilangan Prima")


angka = int(input("Masukkan angka: "))
tampilkan_hasil(angka)
```

```python
def ordinal(angka):
    if angka % 100 == 11 or angka % 100 == 12 or angka % 100 == 13:
        return (angka, "th")
    elif angka % 10 == 1:
        return (angka, "st")
    elif angka % 10 == 2:
        return (angka, "nd")
    elif angka % 10 == 3:
        return (angka, "rd")
    else:
        return (angka, "th")

print("Ordinal Number")

while True:
    print("ketik 0 untuk menghentikan program")
    angka = int(input("masukkan angka: "))

    print(ordinal(angka))

    if angka == 0:
        print("terima kasih telah menggunakan program saya")
        break
```
